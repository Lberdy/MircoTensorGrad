```python
import numpy as np
from MLP import MLP, MicroTensor, SGD, create_Batches
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
```

```python
len(iris['data'])
```

```python
iris['data'] = iris['data'].reshape((150, 4, 1))
```

```python
def transform(array):
    new_array = []
    for element in array:
        if element == 0:
            new_array.append([1, 0, 0])
        elif element == 1:
            new_array.append([0, 1, 0])
        else:
            new_array.append([0, 0, 1])
    return np.array(new_array)
```

```python
iris['target'] = transform(iris['target']).reshape((150, 3, 1))
```

```python
rng = np.random.default_rng(seed=42)
shuffled_indices = rng.permutation(150)
```

```python
iris['data'] = iris['data'][shuffled_indices]
iris['target'] = iris['target'][shuffled_indices]
```

```python
X = iris['data']
Y = iris['target']
```

```python
X_Batches, Y_Batches = create_Batches(X, Y, bacthSize=64)
```

```python
Optimizer = SGD(lr=0.005)
```

```python
model = MLP(
    components=[
        ["Linear", 4, 8],
        ["Linear", 8, 3]
    ],
    LossFunction="CrossEntropy",
    optimizer=Optimizer
)
```

```python
model.training = True
```

```python
epochs = 50
```

```python
for i in range(epochs):
    for X_batch, Y_batch in zip(X_Batches, Y_Batches):
        Xb = MicroTensor(X_batch)
        topo = model(Xb, Y_batch)
        
        topo.backward()
    
        model.update()
    print(f"epoch : {i}, loss : {model(Xb, Y_batch).topo[-1].array}")
```

```python
def accuracy(true : np.ndarray, pred : np.ndarray):
    return np.sum(true.argmax(axis=1) == pred.argmax(axis=1))/len(true)
```

```python
model.training = False
```

```python
true = Y.reshape((Y.shape[0], 3))
pred = model(MicroTensor(X))
pred = pred.reshape(pred.shape[0], 3)
```

```python
accuracy(true, pred)
```
